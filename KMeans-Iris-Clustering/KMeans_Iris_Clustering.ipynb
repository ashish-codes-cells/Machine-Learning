# K-Means Clustering on Iris Dataset

This notebook demonstrates K-Means clustering on the Iris dataset. The workflow follows the lecture approach: data visualization, feature scaling, PCA for dimensionality reduction, the Elbow Method, and K-Means clustering.

## 1. Import Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## 2. Load the Iris Dataset

```python
# Load the Iris dataset
iris = load_iris()

# Separate features and target
X = iris.data
y = iris.target
```

## 3. Visualize the Data

```python
# Visualize two features of the Iris dataset
sns.scatterplot(x=X[:, 0], y=X[:, 2])
plt.title("Iris Dataset: Feature Visualization")
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[2])
plt.show()
```

## 4. Standardize the Features

```python
# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 5. Dimensionality Reduction Using PCA

PCA is used here to reduce the standardized data to two dimensions for visualization and clustering.

```python
# Reduce the data to two principal components
pca = PCA(n_components=2)
pca_data = pca.fit_transform(X_scaled)
```

## 6. Elbow Method

```python
# Calculate WCSS for different numbers of clusters
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=10, n_init=10)
    kmeans.fit(pca_data)
    wcss.append(kmeans.inertia_)
```

```python
# Plot the Elbow Method
sns.lineplot(x=range(1, 11), y=wcss, marker="o")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS")
plt.show()
```

## 7. Apply K-Means Clustering

```python
# Apply K-Means with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=10, n_init=10)
labels = kmeans.fit_predict(pca_data)
```

## 8. Visualize the Clusters

```python
# Plot the clusters and their centroids
sns.scatterplot(x=pca_data[:, 0], y=pca_data[:, 1], hue=labels, palette="deep", legend="full")
sns.scatterplot(
    x=kmeans.cluster_centers_[:, 0],
    y=kmeans.cluster_centers_[:, 1],
    marker="X",
    color="red",
    s=200,
    label="Centroids",
)
plt.title("K-Means Clustering on Iris Dataset")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

## Conclusion

K-Means clustering was applied to the Iris dataset after standardization and PCA-based dimensionality reduction. The Elbow Method was used to examine different values of `k`, and the final clustering was performed using 3 clusters.
